```python
from ipywidgets import Output, GridspecLayout
from IPython.display import display, HTML
from glob import glob
from pathlib import Path
import io
import base64

model_path = "/home/raphael/python/latest/AI23-Deep-Learning/space_invaders/complete/"
filepaths = glob(model_path+"*")
grid = GridspecLayout(15, 3)
players = set()
bracket = (0,4) #(4,8) (8,12) (12,16)

for fp in filepaths[bracket[0]:bracket[1]]:
    path = Path(fp)
    players.add(path.name)

for p, player in enumerate(players):    
    for i, filepath in enumerate(glob(model_path+player+"/*.mp4")):
        out = Output()
        with out:
            print(player)
            video = io.open(filepath, 'r+b').read()
            encoded = base64.b64encode(video)
            display(HTML(data='''
            <video width="320" height="240" controls autoplay muted>
                <source src="data:video/mp4;base64,{0}" type="video/mp4" />
            </video>
            '''.format(encoded.decode('ascii'))))
        grid[p, i] = out

grid
```
